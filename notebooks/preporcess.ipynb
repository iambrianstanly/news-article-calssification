```python
tokens = []
for paragraph in train_set['text'].tolist():
    words_ending_with_dot = ''.join([letter for letter in paragraph.lower() if letter != '.'])
    tokens_ = [words for words in  words_ending_with_dot.split()]

    #remove numbers
    without_numbers = []
    for word in tokens_:
        if re.search(r'\d', word):
            pass
        else:
            without_numbers.append(word)


    # remove
    without_special_characters = []

    for word in without_numbers:
        if re.search(r'[^\w\s]', word):
            pass
        else:
            without_special_characters.append(word)


    tokens.append(without_special_characters)
```

```python
max_length = 0
for sam in tokens:
    len_ = (len(sam))
    if len_ > max_length:
        max_length = len_

max_length
```

```python

tokens_flatten = [word for row in tokens for word in row]
print(len(tokens_flatten))
```

```python
unique_words = (set(tokens_flatten))
```

```python
vocab_size = len(unique_words)
vocab_size
```

```python
vocab = {"<PAD>": 0, "<UNK>": 1}

for i,word in enumerate(unique_words, start=2):
    vocab[word] = i

for word,i in vocab.items():
    print(f"word: {word} | {i}")
```

```python
import json

with open('../configs/vocab.json', 'w') as f:
    json.dump(vocab, f)
```
